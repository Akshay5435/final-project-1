After understanding the dataset, I started the data cleaning process because clean data is important for getting reliable analysis results. I first checked for duplicate records and removed duplicate rows where necessary.I also filtered invalid values such as zero or negative age and purchase amounts. After performing these steps, the dataset was prepared for further analysis and transformation.

```python
import os
import pandas as pd
import numpy as np

# Automatically find the dataset path in Colab
file_path = None
for root, dirs, files in os.walk("/"):
    for file in files:
        if "shoppingbehavior" in file.lower():
            file_path = os.path.join(root, file)
            break

if file_path is None:
    raise FileNotFoundError(
        "File not found! Please upload your dataset using the Files (📁) panel on the left sidebar."
    )

print(f"Loading file from path: {file_path}")

# Load the dataset based on extension
if file_path.endswith('.xlsx') or file_path.endswith('.xls'):
    df = pd.read_excel(file_path)
else:
    df = pd.read_csv(file_path)

print("--- Original Shape ---")
print(df.shape)

# ==============================================================================
# STEP 1: REMOVING DUPLICATES
# ==============================================================================
# Remove exact duplicate rows across all columns
df = df.drop_duplicates()

# Drop duplicates by Customer ID if present
id_cols = [c for c in df.columns if 'id' in c.lower() or 'customer' in c.lower()]
if id_cols:
    df = df.drop_duplicates(subset=[id_cols[0]], keep='first')

print("\n--- Shape After Removing Duplicates ---")
print(df.shape)

# ==============================================================================
# STEP 2: HANDLING MISSING VALUES
# ==============================================================================
df.columns = df.columns.str.strip()

for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype in ['int64', 'float64']:
            # Impute missing numerical values with Median
            df[col] = df[col].fillna(df[col].median())
        else:
            # Impute missing categorical values with Mode
            mode_val = df[col].mode()[0] if not df[col].mode().empty else 'Unknown'
            df[col] = df[col].fillna(mode_val)

# ==============================================================================
# STEP 3: CORRECTING DATA TYPES
# ==============================================================================
# Strip extra whitespace from text fields
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.strip()

# Convert datetime columns if present
date_cols = [c for c in df.columns if 'date' in c.lower() or 'time' in c.lower()]
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Convert numeric columns safely
numeric_keywords = ['age', 'amount', 'price', 'rating', 'review', 'purchase']
for col in df.columns:
    if any(kw in col.lower() for kw in numeric_keywords):
        df[col] = pd.to_numeric(df[col], errors='coerce')

# ==============================================================================
# STEP 4: CREATING DERIVED COLUMNS
# ==============================================================================
# 1. Age Grouping
age_cols = [c for c in df.columns if 'age' in c.lower()]
if age_cols:
    col_age = age_cols[0]
    df['Age_Group'] = pd.cut(
        df[col_age],
        bins=[0, 25, 40, 60, 100],
        labels=['Youth (<25)', 'Young Adult (25-40)', 'Adult (41-60)', 'Senior (60+)']
    )

# 2. High Spender Tag
amt_cols = [c for c in df.columns if 'amount' in c.lower() or 'price' in c.lower() or 'purchase' in c.lower()]
if amt_cols:
    col_amt = amt_cols[0]
    df['Is_High_Spender'] = df[col_amt] >= df[col_amt].quantile(0.75)

# 3. Customer Loyalty Level
prev_cols = [c for c in df.columns if 'previous' in c.lower() or 'history' in c.lower()]
if prev_cols:
    col_prev = prev_cols[0]
    df['Customer_Segment'] = pd.cut(
        df[col_prev],
        bins=[-1, 5, 20, np.inf],
        labels=['Occasional', 'Regular', 'Loyal']
    )

# ==============================================================================
# STEP 5: FILTERING DATA
# ==============================================================================
# Filter out invalid values
if amt_cols:
    df = df[df[amt_cols[0]] > 0]
if age_cols:
    df = df[df[age_cols[0]] > 0]

print("\n=== DATA CLEANING COMPLETE ===")
print("Cleaned Shape:", df.shape)
print("\nMissing values remaining:", df.isnull().sum().sum())
df.head()
```

As part of the cleaning process, I specifically checked the dataset for duplicate records because repeated transactions can affect calculations such as averages, totals and counts. I used Pandas to check and remove exact duplicate rows and also checked records using an ID-related field

I then checked every column in the dataset for missing values because missing information can affect the analysis and charts

```python
# Identify primary category and numerical columns dynamically
amt_col = [c for c in df.columns if 'amount' in c.lower() or 'price' in c.lower() or 'purchase' in c.lower()][0]
gender_col = [c for c in df.columns if 'gender' in c.lower() or 'sex' in c.lower()]
cat_feature = [c for c in df.columns if 'category' in c.lower() or 'item' in c.lower()]

# 1. GroupBy Aggregation: Total & Average Spend by Category
if cat_feature:
    print(f"=== SUMMARY BY {cat_feature[0].upper()} ===")
    category_summary = df.groupby(cat_feature[0])[amt_col].agg(['count', 'mean', 'median', 'sum']).reset_index()
    print(category_summary.sort_values(by='sum', ascending=False))

# 2. Pivot Table: Average Spending across Gender and Age Group / Category
if gender_col and 'Age_Group' in df.columns:
    print("\n=== PIVOT TABLE: Average Spend by Gender & Age Group ===")
    pivot_spending = pd.pivot_table(
        df,
        values=amt_col,
        index='Age_Group',
        columns=gender_col[0],
        aggfunc='mean'
    )
    print(pivot_spending)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set overall design theme
sns.set_theme(style="whitegrid", font_scale=1.0)
palette = sns.color_palette("muted")

# Identify key column names dynamically
amt_col = [c for c in df.columns if any(k in c.lower() for k in ['amount', 'price', 'purchase'])][0]
age_col = [c for c in df.columns if 'age' in c.lower()][0]
cat_col = [c for c in df.columns if any(k in c.lower() for k in ['category', 'item'])][0]
gender_col = [c for c in df.columns if any(k in c.lower() for k in ['gender', 'sex'])][0] if any('gender' in c.lower() or 'sex' in c.lower() for c in df.columns) else None

# Initialize 2x3 Subplot Grid
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Customer Shopping Behavior Analysis Dashboard', fontsize=20, fontweight='bold', y=0.98)

# ------------------------------------------------------------------------------
# 1. HISTOGRAM + KDE: Distribution of Purchase Amounts
# ------------------------------------------------------------------------------
sns.histplot(data=df, x=amt_col, kde=True, ax=axes[0, 0], color='#2b5c8f', bins=25)
axes[0, 0].set_title('1. Distribution of Purchase Amounts', fontweight='bold', pad=10)
axes[0, 0].set_xlabel('Purchase Amount ($)')
axes[0, 0].set_ylabel('Frequency')

# ------------------------------------------------------------------------------
# 2. BAR PLOT: Total Spend by Category
# ------------------------------------------------------------------------------
cat_totals = df.groupby(cat_col)[amt_col].sum().sort_values(ascending=False).head(6)
sns.barplot(x=cat_totals.values, y=cat_totals.index, ax=axes[0, 1], palette='crest')
axes[0, 1].set_title('2. Total Sales Revenue by Category', fontweight='bold', pad=10)
axes[0, 1].set_xlabel('Total Revenue ($)')
axes[0, 1].set_ylabel('Category')

# ------------------------------------------------------------------------------
# 3. BOX PLOT: Spending Distribution across Age Groups
# ------------------------------------------------------------------------------
if 'Age_Group' in df.columns:
    sns.boxplot(data=df, x='Age_Group', y=amt_col, ax=axes[0, 2], palette='Set2')
    axes[0, 2].set_title('3. Purchase Amount Variance by Age Group', fontweight='bold', pad=10)
    axes[0, 2].set_xlabel('Age Group')
    axes[0, 2].set_ylabel('Purchase Amount ($)')

# ------------------------------------------------------------------------------
# 4. SCATTER PLOT: Age vs. Purchase Amount (Grouped by Gender)
# ------------------------------------------------------------------------------
if gender_col:
    sns.scatterplot(data=df, x=age_col, y=amt_col, hue=gender_col, ax=axes[1, 0], alpha=0.7, palette='Set1')
    axes[1, 0].set_title(f'4. Customer Age vs. Spending ({gender_col.capitalize()})', fontweight='bold', pad=10)
    axes[1, 0].set_xlabel('Customer Age')
    axes[1, 0].set_ylabel('Purchase Amount ($)')
    axes[1, 0].legend(title=gender_col.capitalize(), loc='upper right')

# ------------------------------------------------------------------------------
# 5. CORRELATION HEATMAP: Numerical Metrics
# ------------------------------------------------------------------------------
num_df = df.select_dtypes(include=['int64', 'float64'])
corr = num_df.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', ax=axes[1, 1], cbar=True, square=True)
axes[1, 1].set_title('5. Feature Correlation Heatmap', fontweight='bold', pad=10)

# ------------------------------------------------------------------------------
# 6. PIE CHART: Customer Loyalty Segment Share
# ------------------------------------------------------------------------------
if 'Customer_Segment' in df.columns:
    segment_counts = df['Customer_Segment'].value_counts()
    axes[1, 2].pie(
        segment_counts.values,
        labels=segment_counts.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=sns.color_palette("pastel")
    )
    axes[1, 2].set_title('6. Customer Loyalty Distribution', fontweight='bold', pad=10)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# 1. Print all column names to check exact spellings
print("Your columns are:", df.columns.tolist())

# 2. DEFINE YOUR EXACT COLUMNS HERE (Replace strings if needed)
# Example: 'Purchase Amount (USD)', 'Category', 'Age_Group'
amt_col = 'Purchase Amount (USD)' if 'Purchase Amount (USD)' in df.columns else [c for c in df.columns if 'amount' in c.lower() or 'price' in c.lower()][0]
cat_col = 'Category' if 'Category' in df.columns else [c for c in df.columns if 'cat' in c.lower() or 'item' in c.lower()][0]
age_group_col = 'Age_Group' if 'Age_Group' in df.columns else 'Age'

print(f"\nPlotting Treemap using:\n - Hierarchy: [{age_group_col}, {cat_col}]\n - Values: {amt_col}")

# 3. Aggregate data cleanly before sending to Plotly
treemap_df = df.groupby([age_group_col, cat_col])[amt_col].sum().reset_index()

# Ensure types are valid strings and non-zero numbers
treemap_df[age_group_col] = treemap_df[age_group_col].astype(str)
treemap_df[cat_col] = treemap_df[cat_col].astype(str)
treemap_df = treemap_df[treemap_df[amt_col] > 0]

# 4. Render Treemap
fig_treemap = px.treemap(
    treemap_df,
    path=[age_group_col, cat_col],
    values=amt_col,
    color=amt_col,
    color_continuous_scale='Viridis',
    title=f"<b>Revenue Contribution by {age_group_col} & {cat_col} Hierarchy</b>",
    template="plotly_white"
)

fig_treemap.update_layout(title_x=0.5, margin=dict(t=50, l=10, r=10, b=10))
fig_treemap.show()
```

The treemap clearly showed which age groups and categories contributed most to overall spending.

Larger rectangles represented categories with higher total purchases, while smaller ones showed less significant contributions.

This visualization provided an easy way to compare spending patterns across both product categories and customer demographics in one chart.
